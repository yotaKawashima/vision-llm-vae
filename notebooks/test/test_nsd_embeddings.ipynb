```python
import pandas as pd
from torch.utils.data import Dataset
import matplotlib.pyplot as plt
import numpy as np
import json
import torch
```

```python
meta_data = "/workspace/data/preprocessed/text_embeddings/meta_text_embeddings_mean_nsdDoerig.json"
with open(meta_data, "r", encoding="utf-8") as f:
    meta = json.load(f)
```

```python
text_embeddings_data = "/workspace/data/preprocessed/text_embeddings/text_embeddings_mean_nsdDoerig.pt"
text_embeddings = torch.load(text_embeddings_data)

text_embeddings.shape
```

```python
# check distribution of mean value for each dim
embs_np = text_embeddings.numpy()
means = np.mean(embs_np, axis=0)  # mean of each embedding
print(means.shape)
plt.hist(means, bins=50)
plt.title("Histogram of Means of Text Embeddings (each dimension)")
plt.xlabel("Mean Value")
plt.ylabel("Frequency")
plt.show()  

# check L2 norm of each embedding
l2_norm = np.linalg.norm(embs_np, axis=1)  
plt.scatter(range(len(l2_norm)), l2_norm)
plt.ylim(0.9, 1.1)
plt.title("Scatter Plot of L2 Norms of Text Embeddings (each text)")
plt.xlabel("Index")
plt.ylabel("L2 Norm Value")
plt.show()
```
