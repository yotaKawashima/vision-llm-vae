```python
import pandas as pd
from torch.utils.data import Dataset
import os 
import matplotlib.pyplot as plt
import numpy as np
```

```python
stimulus_info_path = "/workspace/data/nsd/nsd_stim/nsd_stim_info_merged.pkl"
stimulus_info = pd.read_pickle(stimulus_info_path)

special515_data_path = "/workspace/data/nsd/nsd_stim/nsd_stim_info_special515.pkl"

print(f"Number of stimuli: {len(stimulus_info)}")
print(f"Unique coco splits: {stimulus_info.cocoSplit.unique()}")
print(f"Columns in stimulus info: {stimulus_info.columns}")
print("----")
stimulus_info[0:2]
```

```python
# 515 stimuli shared across all participants, within valid sessions.

num_trials = 750
num_sessions = {'1': 40, '2': 40, '3': 32, '4': 30, '5': 40, '6': 32, '7': 40, '8': 30}
max_trials = {subject: num_trials * sessions for subject, sessions in num_sessions.items()}
is_special515 = (stimulus_info['shared1000'] == True)

for participant_id, limit in max_trials.items():
    col = f'subject{participant_id}_rep2'
    is_special515 &= (stimulus_info[col] > 0) & (stimulus_info[col] <= limit)

special515_df = stimulus_info[is_special515]
is_not_special515 = (~is_special515)

assert len(special515_df) == 515, f"Expected 515 stimuli in the special set, but found {len(special515_df)}"

special515_df[['cocoId', 'cocoSplit', 'nsdId']]

# save the special 515 stimuli info to a pkl file
special515_df.to_pickle(special515_data_path)
```

```python
# for each subject, check how many stimuli are not in the special 515 set, but are still within the valid session limits.
for participant_id, limit in max_trials.items():
    present_trials = (stimulus_info[f'subject{participant_id}_rep2'] > 0)
    valid_trials = (stimulus_info[f'subject{participant_id}_rep2'] <= limit)
    is_not_special515_this_participant = is_not_special515 & present_trials & valid_trials
    print(f"{participant_id}, Not Special 515 sum: {(is_not_special515_this_participant.sum().item())}")

    this_participant_df = stimulus_info[is_not_special515_this_participant][['cocoId', 'cocoSplit', 'nsdId']]

    this_participant_df.to_pickle(f"/workspace/data/nsd/nsd_stim/subj{int(participant_id):02d}_nsd_stim_info_NOTspecial515.pkl")
```

```python
from typing import Optional 

class NSDStimulusDataset(Dataset):
    """
    NSDStimulusDataset
    -----------------------------

    PyTorch Dataset that provides images from the NSD Stimulus dataset.

    """

    def __init__(
        self,
        nsd_stimulus_info_path="/workspace/data/nsd/nsd_stim_info_merged.pkl",
        img_transform=None,
    ):

        # image
        self.coco_train_img_dir = "/workspace/data/coco/train2017"
        self.coco_val_img_dir = "/workspace/data/coco/val2017"

        # load stimulus info
        self.stimulus_info = pd.read_pickle(nsd_stimulus_info_path)
        self.stimulus_info_shared1000 = self.stimulus_info[
            self.stimulus_info.shared1000
        ]

        self.img_transform = img_transform

    def __len__(self):
        return len(self.stimulus_info_shared1000)

    def __getitem__(
            self, 
            idx: int, 
            subject: Optional[int] = None, 
            ):

        # image file path
        if self.stimulus_info_shared1000.iloc[idx]["cocoSplit"] == "train2017":
            img_dir = self.coco_train_img_dir
        elif self.stimulus_info_shared1000.iloc[idx]["cocoSplit"] == "val2017":
            img_dir = self.coco_val_img_dir
        else:
            raise ValueError(
                f"Invalid cocoSplit value: {self.stimulus_info_shared1000.iloc[idx]['cocoSplit']}"
            )

        if subject is None

        image_id = self.stimulus_info_shared1000.iloc[idx]["cocoId"]
        img_path = os.path.join(img_dir, f"{image_id:012d}.jpg")
        print(img_path)
        # image = Image.open(img_path).convert("RGB")
        # image = self.img_transform(image)
        image = None
        return {
            "image": image,
            "image_id": image_id.item(),
        }
```
